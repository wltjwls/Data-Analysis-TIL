```python
#판다스 프로그램 불러오기 
import pandas as pd
```

```python
#공항 데이터 만들기
airport_data ={
    '공항명' : ['김포', '제주', '김해', '대구', '인천'],
    '일일이용객' : [75000, 82000, 45000, 12000, 9000],
    '운항편수' : [400, 450, 250, 80, 60]

}
```

```python
#데이터를 표로 변환하기
df = pd.DataFrame(airport_data)
```

```python
#표 확인하기
df
```

```python
#표의 가장 위쪽 3줄 미리보기
df.head(3)
```

```python
#표의 전반적인 정보(줄 개수, 데이터 종류 등) 확인하기
df.info()
```

```python
#일일이용객이 5만 명이 넘는 공항만 필터링하기
busy_airports = df[df['일일이용객']>=50000]
```

```python
#결과 확인
busy_airports
```
