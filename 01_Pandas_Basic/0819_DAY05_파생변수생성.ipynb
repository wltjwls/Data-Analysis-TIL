```python
import pandas as pd

# 공항의 일일 운항 및 이용객 데이터
data = {
    '공항명': ['김포', '제주', '김해', '청주'],
    '일일운항편수': [400, 480, 250, 50],
    '일일이용객': [75000, 84000, 45000, 6000]
}
df = pd.DataFrame(data)

df
```

```python
#'일일이용객'을 '일일운항편수'로 나눠서 새로운 열 '편당_탑승객' 만들기
df['편당_탑승객']=df['일일이용객']/df['일일운항편수']

#반올림(round) 처리
df['편당_탑승객']=df['편당_탑승객'].round(0)

#결과물 확인
df
```

### 파생 변수(새로운 열) 만들기

#### df['새로운 열'] = 계산식 
* 사칙연산이나 함수를 통해 데이터를 만들어 냄 
* ex. (매출액)-(원가)='[순이익]' 
* **round(숫자)** 계산 결과가 지저분하게 나올 때 이용
