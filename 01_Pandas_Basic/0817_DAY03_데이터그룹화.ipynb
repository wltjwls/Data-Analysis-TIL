```python
import pandas as pd

# Day 2에서 결측치를 모두 정제 완료한 깨끗한 데이터
data = {
    '공항명': ['김포', '제주', '김해', '청주', '무안', '대구'],
    '일일이용객': [75000, 82000, 45000, 43400, 3000, 12000], 
    '지연건수': [12, 5, 8, 2, 0, 0], 
    '혼잡도': ['높음', '높음', '보통', '낮음', '낮음', '보통']
}
df = pd.DataFrame(data)

df
```

```python
#'혼잡도'를 기준으로 묶어서, 숫자 데이터들의 평균 구하기
df_grouped = df.groupby('혼잡도').mean(numeric_only=True)

#결과 확인 (소수점 첫째 자리까지만 반올림)
df_grouped.round(1)
```

### 통계 내기 필수 함수 

#### groupby() 
* 데이터를 그룹 지어주는 역할 
* 자주 쓰는 통계 함수
    * .mean() 평균
    * .sum() 합계
    * .count() 개수
    * .max()/.min() 최댓값/최솟값
    * ex.'df.groupby('항목').sum()

##### numeric_only=True 숫자로 된 데이터만 계산
